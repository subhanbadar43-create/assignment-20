```python
def calculatefare(distance, peak=False):
    if distance <= 5:
        fare = 100
    elif distance <= 10:
        extrakm = distance - 5
        fare = 100 + extrakm * 30
    else:
        extrakm = distance - 10
        fare = 100 + (5 * 30) + extrakm * 25
    if peak == True:
        extracharge = fare * 20 / 100
        fare = fare + extracharge
    print("Distance =", distance, "km")
    print("Peak Hour =", peak)
    print("Total Fare = Rs.", fare)
    print("------------------------")
    return fare
calculatefare(3)
calculatefare(8)
calculatefare(15)
calculatefare(15, True)
```
